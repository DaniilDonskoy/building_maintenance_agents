```python
import sys
import os
from pathlib import Path

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
PARENT = PROJECT_ROOT.parent
if str(PARENT) not in sys.path:
    sys.path.insert(0, str(PARENT))
```

```python
from building_maintenance_agents.house_graph.samples import House16Factory
from building_maintenance_agents.incident_simulator import IncidentSimulator
```

```python
house = House16Factory.build()
print(f"House created: {len(house.nodes)} nodes, {len(house.edges)} edges")

simulator = IncidentSimulator(
		house,
		base_incident_probability=0.05,
		random_seed=42,
		enable_spread=True
)
```

Run simulation:

```python
for step in range(3000):
		results = simulator.step()
		
		if step % 10 == 0 or results["new_incidents"] or results["resolved_incidents"]:
				print(f"\nStep {step}:")
				print(f"  Active incidents: {results['active_count']}")
				
				if results["new_incidents"]:
						print(f"  New incidents: {len(results['new_incidents'])}")
						for inc in results["new_incidents"][:3]:
								print(f"    - {inc['type']} at {inc['location']} (severity: {inc['severity']:.2f})")
				
				if results["resolved_incidents"]:
						print(f"  Resolved incidents: {len(results['resolved_incidents'])}")
```

Statistics:

```python
stats = simulator.get_statistics()
for key, value in stats.items():
		if key != "incidents_by_type":
				print(f"{key}: {value}")

print("\nIncidents by type:")
for inc_type, count in stats["incidents_by_type"].items():
		print(f"  {inc_type}: {count}")
```

In tensors:

```python
tensors = simulator.get_incident_tensors()
for name, tensor in tensors.items():
		print(f"{name}: shape={tensor.shape}, sum={tensor.sum().item():.2f}")
```
